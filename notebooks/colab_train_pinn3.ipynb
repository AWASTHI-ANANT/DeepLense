# Train PINN v3 on Colab (T4)

PINN v3 (`LensingPINNv3` in `deeplense/models/pinn_lens.py`): an SIE + shear lens, Sérsic source and PSF
matching the DeepLenseSim generator, plus a U-Net substructure decoder. Three runs:

| run | what it tests |
|---|---|
| `pinn3_physonly_lambda0.1` | **key experiment**: classify from physics outputs only (residual, κ_sub, lens params), detached from the CE loss |
| `pinn3_lambda0.1` | full model (backbone + physics features) |
| `pinn3_lambda0` | full model, Poisson loss off (ablation) |

Defaults for v3: κ_sub sparsity 0.1 and a minimum-deflection prior (weight 30) on ∇δψ. Local tests showed
that without them the perturbation exploits two degeneracies: mass in empty sky, which doesn't change the
image, and massless harmonic modes, which satisfy the Poisson term with κ_sub = 0.

**Before running:** *Runtime → Change runtime type → T4 GPU*.

GPU quota runs out. You can't fully prevent that, so this notebook makes it harmless instead:

## 3. Train
Each command prints one line per epoch (time, losses, `theta_e_px`, val AUC), writes a log to Drive,
and resumes automatically if a checkpoint exists. A finished run (with `metrics.json`) is skipped.

```python
import subprocess, sys

def train(lam, physics_only=False):
    run = f"pinn3{'_physonly' if physics_only else ''}_lambda{lam:g}"
    if os.path.exists(f'{OUT}/{run}/metrics.json'):
        print(f'{run}: already finished, skipping'); return
    cmd = [sys.executable, '-u', 'scripts/train.py', '--model', 'pinn3', '--lambda-poisson', str(lam),
           '--data-root', DATA, '--out-dir', OUT, '--batch-size', '64', '--num-workers', '2',
           '--amp', '--resume'] + (['--physics-only'] if physics_only else [])
    with open(f'{OUT}/{run}.log', 'a') as log:
        proc = subprocess.Popen(cmd, cwd=REPO, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
        for line in proc.stdout:            # stream each epoch line to the cell and the Drive log
            if 'HF_TOKEN' not in line:
                print(line, end=''); log.write(line); log.flush()
    if proc.wait() != 0:
        raise RuntimeError(f'{run} failed (exit {proc.returncode}); see the output above')

train(0.1, physics_only=True)    # key experiment: physics-only classifier
```

```python
train(0.1)    # full model, Poisson loss on
```

```python
train(0)      # full model, Poisson loss off (ablation)
```

## 4. Results

```python
import json
for run in ['pinn3_physonly_lambda0.1', 'pinn3_lambda0.1', 'pinn3_lambda0']:
    f = f'{OUT}/{run}/metrics.json'
    if os.path.exists(f):
        t = json.load(open(f))['test']
        print(f"{run:18s} test acc {t['accuracy']:.4f}  macro AUC {t['macro_auc']:.4f}  ",
              {k.split()[0]: round(v, 4) for k, v in t['auc_per_class'].items()})
```
